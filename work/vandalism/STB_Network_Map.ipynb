```python
from mpl_toolkits.basemap import Basemap

import matplotlib.pyplot as plt

# Indian Ocean domain boundaries
lon_min, lon_max = 45, 105
lat_min, lat_max = -6, 35

# Tsunami buoy station locations (example coordinates)
stations = {
    'STB-01': (6, 88),
    'STB-02': (20, 63),
    'STB-03': (3, 89),
    'STB-04': (10, 85),
    'STB-05': (12, 86)
}

plt.figure(figsize=(10, 8))
m = Basemap(projection='merc',
            llcrnrlat=lat_min, urcrnrlat=lat_max,
            llcrnrlon=lon_min, urcrnrlon=lon_max,
            resolution='i')

m.drawcoastlines()
m.drawcountries()
m.drawmapboundary(fill_color='aqua')
m.fillcontinents(color='lightgray', lake_color='aqua')
m.drawparallels(range(lat_min, lat_max+1, 10), labels=[1,0,0,0])
m.drawmeridians(range(lon_min, lon_max+1, 20), labels=[0,0,0,1])

# Plot stations
for label, (lon, lat) in stations.items():
    x, y = m(lon, lat)
    m.plot(x, y, 'ro', markersize=8)
    plt.text(x, y+100000, label, fontsize=12, ha='center', va='bottom', color='blue')

plt.title('Indian Ocean Tsunami Buoy Network')
plt.tight_layout()
plt.show()
```
