```python
import matplotlib.pyplot as plt

station_ids = ['STB-01', 'STB-02', 'STB-03', 'STB-04', 'STB-05']
deployments = [7, 7, 2, 1, 2]  # example data for deployments
vandalism = [2, 3, 2, 1, 1]    # example data for vandalism incidents

x = range(len(station_ids))
width = 0.35

plt.figure(figsize=(8, 5))
plt.bar([i - width/2 for i in x], deployments, width=width, label='Deployments')
plt.bar([i + width/2 for i in x], vandalism, width=width, label='Vandalism Incidents')

plt.xticks(x, station_ids)
plt.xlabel('Tsunami Buoy Station ID')
plt.ylabel('Count')
plt.title('Deployments and Vandalism Incidents per Station')
plt.ylim(0, max(deployments) + 2)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Calculate percentage of vandalism incidents at each station
percent_vandalism = [v / d * 100 if d != 0 else 0 for v, d in zip(vandalism, deployments)]

# Display the percentage for each station
for station, percent in zip(station_ids, percent_vandalism):
    print(f"{station}: {percent:.1f}% vandalism incidents")
```

```python
plt.figure(figsize=(8, 5))
plt.bar([i - width/2 for i in x], deployments, width=width, label='Deployments')
plt.bar([i + width/2 for i in x], vandalism, width=width, label='Vandalism Incidents')

# Annotate each station with percentage of vandalism incidents
for i, percent in enumerate(percent_vandalism):
    plt.text(i + width/2, vandalism[i] + 0.2, f"{percent:.1f}%", ha='center', va='bottom', fontsize=10, color='red')

plt.xticks(x, station_ids)
plt.xlabel('Tsunami Buoy Station ID')
plt.ylabel('Count')
plt.title('Deployments and Vandalism Incidents per Station')
plt.ylim(0, max(deployments) + 2)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Plot composite bars: bottom is deployments minus vandalism, top is vandalism
non_vandalism = [d - v for d, v in zip(deployments, vandalism)]
plt.bar(x, non_vandalism, width=width, label='Non-Vandalism Deployments', color='skyblue')
plt.bar(x, vandalism, width=width, bottom=non_vandalism, label='Vandalism Incidents', color='salmon')

# Annotate each bar with percentage of vandalism incidents
for idx, (total, percent) in enumerate(zip(deployments, percent_vandalism)):
    plt.text(idx, total + 0.2, f"{percent:.1f}%", ha='center', va='bottom', fontsize=10, color='red')

plt.xticks(x, station_ids)
plt.xlabel('Tsunami Buoy Station ID')
plt.ylabel('Total Deployments')
plt.title('Composite Bar: Deployments and Vandalism Incidents per Station')
plt.ylim(0, max(deployments) + 3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
total_deployments = sum(deployments)
total_vandalism = sum(vandalism)
overall_percent_vandalism = (total_vandalism / total_deployments) * 100 if total_deployments != 0 else 0

plt.figure(figsize=(5, 5))
plt.bar(['Deployments', 'Vandalism Incidents'], [total_deployments, total_vandalism], color=['skyblue', 'salmon'])
plt.ylabel('Total Count')
plt.title('Total Deployments vs Vandalism Incidents (All Stations)')

# Annotate vandalism bar with percentage
plt.text(1, total_vandalism + 0.2, f"{overall_percent_vandalism:.1f}%", ha='center', va='bottom', fontsize=12, color='red')

plt.ylim(0, total_deployments + 3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Plot single bar for each station: deployments stacked with vandalism
plt.bar(x, deployments, width=width, label='Deployments', color='skyblue')
plt.bar(x, vandalism, width=width, bottom=deployments, label='Vandalism Incidents', color='salmon')

# Annotate each bar with total (deployments + vandalism)
for idx in x:
    total = deployments[idx] + vandalism[idx]
    plt.text(idx, total + 0.2, f"{total}", ha='center', va='bottom', fontsize=10, color='black')

plt.xticks(x, station_ids)
plt.xlabel('Tsunami Buoy Station ID')
plt.ylabel('Count')
plt.title('Deployments and Vandalism Incidents per Station (Stacked Bar)')
plt.ylim(0, max([d + v for d, v in zip(deployments, vandalism)]) + 3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Plot composite bars: bottom is deployments minus vandalism, top is vandalism
non_vandalism = [d - v for d, v in zip(deployments, vandalism)]
plt.bar(x, non_vandalism, width=width, label='Non-Vandalized', color='skyblue', alpha = 0.8)
plt.bar(x, vandalism, width=width, bottom=non_vandalism, label='Vandalism Incidents', color='salmon', alpha = 0.8)

# Annotate each bar with number of deployments at the top
for idx in x:
    plt.text(idx, deployments[idx] + 0.2, f"{deployments[idx]}", ha='center', va='bottom', fontsize=10, color='black')
    # Annotate vandalism portion with number of incidents
    if vandalism[idx] > 0:
        plt.text(idx, non_vandalism[idx] + vandalism[idx]/2, f"{vandalism[idx]}", ha='center', va='center', fontsize=10, color='red')
    if non_vandalism[idx] > 0:
        plt.text(idx, non_vandalism[idx] - vandalism[idx]/2, f"{non_vandalism[idx]}", ha='center', va='center', fontsize=10, color='blue')
plt.xticks(x, station_ids)
plt.xlabel('Tsunami Buoy Station ID')
plt.ylabel('Total Deployments')
plt.title('Deployments and Vandalism Incidents per Station')
plt.ylim(0, max(deployments) + 2)
plt.legend()
plt.tight_layout()
plt.show()
```
